```
import pandas as pd
import numpy as np

# Ensure datetime
df["CONNECTIONTIME_UTC"] = pd.to_datetime(df["CONNECTIONTIME_UTC"])
df["DISCONNECTTIME_UTC"] = pd.to_datetime(df["DISCONNECTTIME_UTC"])

# Create list of hourly timestamps each session covers
df["active_hours"] = df.apply(
    lambda row: pd.date_range(
        start=row["CONNECTIONTIME_UTC"].floor("H"),
        end=row["DISCONNECTTIME_UTC"].floor("H"),
        freq="H"
    ),
    axis=1
)

df_expanded = df.explode("active_hours")


df_expanded["hour_of_day"] = df_expanded["active_hours"].dt.hour


hourly_usage = (
    df_expanded.groupby("hour_of_day")
    .size()
    .reset_index(name="active_sessions")
)
```

```
hourly_df["location"] = "Location_A"
hourly_df_jpl["location"] = "JPL"
hourly_df2["location"] = "Location_B"
```

```
combined_df = pd.concat(
    [hourly_df, hourly_df_jpl, hourly_df2],
    ignore_index=True
)
```

```
combined_df["normalized_usage"] = (
    combined_df.groupby("location")["active_sessions"]
    .transform(lambda x: x / x.sum())
)
```

```
import matplotlib.pyplot as plt
import numpy as np

# Pivot table for grouped plotting
pivot_df = combined_df.pivot(
    index="hour_of_day",
    columns="location",
    values="active_sessions"  # or "normalized_usage"
)

hours = pivot_df.index
locations = pivot_df.columns
x = np.arange(len(hours))
width = 0.25

plt.figure()

for i, loc in enumerate(locations):
    plt.bar(
        x + i * width,
        pivot_df[loc],
        width=width,
        label=loc
    )

plt.xlabel("Hour of Day (UTC)")
plt.ylabel("Active Charging Sessions")
plt.title("EV Charging Usage Comparison by Location")
plt.xticks(x + width, hours)
plt.legend()
plt.show()
```
